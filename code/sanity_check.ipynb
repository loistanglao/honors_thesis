# Check if data processed through pipeline is the same as data processed through NNS EDA notebook

```python
# check working directory
import os
print(os.getcwd())
```

```python
# load in the data
notebook_data = pd.read_csv('../data/processed/only_nns_18_19_21_cleaned.csv')
pipeline_data = pd.read_csv('../data/processed/pipeline_18_19_21.csv')
```

```python
# check if both dataframes are the same
notebook_data.equals(pipeline_data)
```

```python
# check shape first
print(notebook_data.shape, pipeline_data.shape)

# then find differing values
diff = notebook_data.compare(pipeline_data)
print(diff)
```

```python
for col in notebook_data.columns:
    if notebook_data[col].dtype == object:  # skip string columns
        if not notebook_data[col].equals(pipeline_data[col]):
            print(f"REAL difference in column: {col}")
        else:
            print(f"OK: {col}")
    else:
        mask = notebook_data[col].notna() & pipeline_data[col].notna()
        if not np.allclose(notebook_data[col][mask], pipeline_data[col][mask], equal_nan=True):
            print(f"REAL difference in column: {col}")
        else:
            print(f"OK: {col}")
```
